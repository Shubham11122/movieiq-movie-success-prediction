```python
import ast
```

```python
def extract_genres(genre_string):
    try:
        genres = ast.literal_eval(genre_string)
        return ", ".join([genre["name"] for genre in genres])
    except:
        return None
```

```python
[{'id':28,'name':'Action'},
 {'id':12,'name':'Adventure'}]
```

```python
df["genres"] = df["genres"].apply(extract_genres)
```

```python
df[["title", "genres"]].head()
```

```python
df["primary_genre"] = df["genres"].str.split(",").str[0].str.strip()
```

```python
df[["genres", "primary_genre"]].head(10)
```

```python
# Success = 1 if Revenue > Budget else 0
df["success"] = (df["revenue"] > df["budget"]).astype(int)

df["success"].value_counts()
```

```python
df["profit"] = df["revenue"] - df["budget"]

df[["budget","revenue","profit"]].head()
```

```python
df["roi"] = (df["profit"] / df["budget"]) * 100

df[["budget","profit","roi"]].head()
```
